### 1.2 Dota 2 en 5 minutos (para quien nunca jugó)

**Dota 2** es un videojuego de estrategia por equipos: **5 contra 5**. Un equipo juega del lado **Radiant** y el otro del lado **Dire**.
Gana el equipo que destruye la base principal del rival. Existe una escena profesional con ligas, torneos y premios millonarios.

Cada partida profesional tiene dos fases:

1. **Draft (antes de jugar).** Los capitanes de ambos equipos se turnan para **prohibir** (*ban*) y **elegir** (*pick*) personajes llamados **héroes**. Un héroe elegido por un equipo ya no lo puede usar el otro.
2. **Partida.** Los 10 héroes elegidos se enfrentan. El resultado se registra como victoria de Radiant o de Dire.

| Término | Qué significa | Si fuera fútbol... |
|---|---|---|
| **Héroe** | Personaje con habilidades propias. Hay más de 120 | Un jugador del plantel con un estilo propio |
| **Pick** | Elegir un héroe para tu equipo | Poner a un jugador en tu once titular |
| **Ban** | Prohibir un héroe para que nadie lo use | Vetar a la estrella rival antes del partido |
| **Draft** | La secuencia completa de bans y picks | La pizarra táctica antes del partido |
| **Captains Mode** | El formato de draft que se usa en torneos profesionales | El reglamento oficial del torneo |
| **Radiant / Dire** | Los dos lados del mapa | Local / visitante |
| **Parche** (7.40, 7.41…) | Actualización del juego que cambia el poder de los héroes | Un cambio de reglamento entre temporadas |
| **Meta** | Los héroes y estrategias que "están de moda" porque funcionan en el parche actual | El sistema táctico que todos copian esa temporada |
| **Pick rate / Ban rate** | % de partidas en que un héroe fue elegido / prohibido | % de partidos en que un jugador fue titular / vetado |
| **Contest rate** | Pick rate + ban rate: cuánto "importa" un héroe en el draft | Qué tan disputado es un jugador en el mercado |
| **Win rate** | % de veces que ganó el equipo que eligió ese héroe | % de victorias cuando ese jugador es titular |
| **Héroes efectivos** | "Es como si se usaran N héroes por igual" (medida de diversidad) | Cuántos jugadores rotan realmente en el equipo |

> **Por qué importa:** si en un parche se usan pocos héroes (meta concentrado), un analista puede preparar el draft enfocado en ellos.
> Si el meta es diverso, la preparación debe ser más amplia. Esa es la decisión que este proyecto quiere apoyar.

> El glosario completo para la exposición está en `docs/glosario_dota2.md`.

## 6. Perfil descriptivo

### 6.1 Variables derivadas: de códigos a información legible

````python
meta["fecha"] = pd.to_datetime(meta["start_date_time"], errors="coerce")
meta["mes"] = meta["fecha"].dt.to_period("M").astype("string")
meta["duracion_min"] = meta["duration"] / 60
meta["primera_sangre_min"] = meta["first_blood_time"] / 60
meta["ganador"] = meta["radiant_win"].map({True: "Radiant", False: "Dire"})
meta["modo_juego"] = meta["game_mode"].map(config.MODOS_JUEGO).fillna("Otro")
meta["tipo_lobby"] = meta["lobby_type"].map(config.TIPOS_LOBBY).fillna("Otro")

# Parche: número interno -> versión legible, ordenada cronológicamente
# Ojo: main_metadata YA tiene una columna "version" (versión del procesamiento de datos,
# p. ej. 22.0), que no es el parche. Por eso la columna legible se llama "parche".
orden_parches = parches.sort_values("patch")["parche"].tolist()
meta = meta.merge(parches[["patch", "parche"]], on="patch", how="left", validate="many_to_one")
meta["parche"] = pd.Categorical(meta["parche"], categories=orden_parches, ordered=True)

meta[["match_id", "fecha", "parche", "duracion_min", "ganador", "modo_juego", "tipo_lobby"]].head()
````

### 6.2 `describe()` explicado

| Estadístico | Qué responde |
|---|---|
| `count` | cuántos valores no nulos hay (compáralo con el total → incompletitud) |
| `mean` | promedio; sensible a valores extremos |
| `std` | dispersión alrededor del promedio |
| `min` / `max` | extremos; ayudan a detectar valores imposibles |
| `25%`, `50%`, `75%` | cuartiles; el 50 % es la **mediana**, robusta ante extremos |
| `skew` | asimetría: > 0.5 cola a la derecha; < −0.5 cola a la izquierda |

> No se interpretan promedios de `match_id`, `leagueid`, `patch` o `hero_id`: son **identificadores o códigos**, no cantidades.

````python
cols_num = ["duracion_min", "primera_sangre_min", "radiant_score", "dire_score", "human_players"]
perfil = meta[cols_num].describe(percentiles=[.01, .25, .5, .75, .99]).T
perfil["nulos"] = meta[cols_num].isna().sum()
perfil["skew"] = meta[cols_num].skew()
display(perfil.round(2))

print("LECTURA GUIADA")
for col in cols_num:
    s = meta[col].dropna()
    rel = "mayor" if s.mean() > s.median() else "menor o igual"
    forma = "cola a la derecha" if s.skew() > 0.5 else "cola a la izquierda" if s.skew() < -0.5 else "aprox. simétrica"
    print(f"• {col}: N={len(s):,}; media={s.mean():.2f}; mediana={s.median():.2f}; "
          f"P1={s.quantile(.01):.2f}; P99={s.quantile(.99):.2f}. Media {rel} que mediana; {forma}.")
````

````python
# Variables categóricas: ¿cuántas categorías hay y cuál domina?
cols_cat = ["parche", "ganador", "modo_juego", "tipo_lobby", "leagueid", "region"]
display(meta[cols_cat].astype("string").describe().T)

for col in ["modo_juego", "tipo_lobby", "ganador"]:
    print(f"\n{col}:"); print(meta[col].value_counts(dropna=False).to_string())
````

````python
# ¿Cuántas partidas hay por parche y por mes? Un parche con pocas partidas pesa poco en las conclusiones.
fig, axes = plt.subplots(1, 2, figsize=(15, 4.5))

por_parche = meta["parche"].value_counts().sort_index()
por_parche = por_parche[por_parche > 0]
axes[0].bar(por_parche.index.astype(str), por_parche.values, color=COLORES["steel"])
axes[0].set(title="Partidas profesionales por parche", xlabel="Parche", ylabel="Partidas")
for i, v in enumerate(por_parche.values):
    axes[0].text(i, v, f"{v:,}", ha="center", va="bottom", fontsize=9)

por_mes = meta.groupby("mes").size()
axes[1].plot(por_mes.index, por_mes.values, marker="o", color=COLORES["navy"])
axes[1].set(title="Partidas por mes", xlabel="Mes", ylabel="Partidas")
axes[1].tick_params(axis="x", rotation=60)

plt.tight_layout()
plt.savefig(config.FIGURAS / "partidas_por_parche_y_mes.png", dpi=150, bbox_inches="tight")
plt.show()
````

````python
# Perfil del draft: ¿cuántas acciones, picks y bans hay por partida?
acciones = pb.groupby(config.PB_PARTIDA).agg(acciones=(config.PB_ES_PICK, "size"),
                                              picks=(config.PB_ES_PICK, "sum"))
acciones["bans"] = acciones["acciones"] - acciones["picks"]
display(acciones.describe().T.round(2))
print("Combinaciones más frecuentes (acciones, picks, bans):")
print(acciones.value_counts().head(5).to_string())
````

**Interpretación del perfil** (dato → significado → consecuencia)

- **Duración:** la partida típica dura **31.9 min** (mediana). La media es 33.9 min porque hay una cola de partidas largas: asimetría 1.18, y el 1 % más largo supera los 66 min. En el otro extremo, el 1 % más corto dura menos de 15.7 min y el mínimo es 6.1 min, algo compatible con rendiciones tempranas (se marcan en 8.5). → Para describir el ritmo del meta conviene usar la **mediana**.
- **Primera sangre:** mediana 1.8 min. Hay valores de 0 segundos (al menos el 1 %) que pueden significar "sin registro" más que una muerte en el segundo 0, así que se interpretan con cautela.
- **Marcador:** Radiant y Dire anotan unas 30 eliminaciones por partida, con distribuciones simétricas.

> Para cada regla: justifica el umbral con conocimiento del juego y escribe el impacto **en términos de tu pregunta**.
> **No se borra nada aquí**: solo se mide. La exclusión, si corresponde, ocurre en Silver (sección 10) con justificación.

````python
cal = EvaluadorCalidad()
heroes_validos = set(heroes["hero_id"])
````

### 8.1 Completitud

````python
# EJEMPLO COMPLETO (modelo para las demás reglas)
cal.evaluar(
    regla="patch no nulo",
    dimension="Completitud",
    tabla="main_metadata",
    cumple=meta["patch"].notna(),
    impacto="Sin parche, la partida no puede asignarse a una versión: queda fuera de la comparación parche a parche",
    responsable="Mantenedor del dataset (o el equipo: imputar por fecha de lanzamiento)",
)

# Equipos identificados: ambos lados deben tener id de equipo
cal.evaluar(
    regla="equipos identificados (Radiant y Dire)",
    dimension="Completitud",
    tabla="main_metadata",
    cumple=meta["radiant_team_id"].notna() & meta["dire_team_id"].notna(),
    impacto="No afecta al draft (los picks existen igual), pero impide analizar el meta por equipo",
    responsable="Mantenedor del dataset / OpenDota (registro de equipos)",
)

# Cada acción de draft debe indicar qué héroe se eligió o prohibió
cal.evaluar(
    regla="hero_id no nulo en picks_bans",
    dimension="Completitud",
    tabla="picks_bans",
    cumple=pb[config.PB_HEROE].notna(),
    impacto="Una acción sin héroe no puede contarse en pick rate, ban rate ni contest rate",
    responsable="Mantenedor del dataset",
)
````

### 8.2 Unicidad

````python
cal.evaluar(
    regla="match_id único en main_metadata",
    dimension="Unicidad",
    tabla="main_metadata",
    cumple=~meta["match_id"].duplicated(keep=False),
    impacto="Una partida repetida cuenta doble en pick rate y win rate",
    responsable="Equipo de análisis (deduplicar en Silver)",
)

# Cada turno del draft (order) aparece una sola vez por partida
cal.evaluar(
    regla="(match_id, order) único en picks_bans",
    dimension="Unicidad",
    tabla="picks_bans",
    cumple=~pb.duplicated(subset=[config.PB_PARTIDA, config.PB_ORDEN], keep=False),
    impacto="Un turno repetido duplica un pick o un ban e infla las tasas de ese héroe",
    responsable="Equipo de análisis (deduplicar en Silver)",
)
````

### 8.3 Validez

````python
# El héroe debe existir en el catálogo de OpenDota
cal.evaluar(
    regla="hero_id existe en el catálogo de héroes",
    dimension="Validez",
    tabla="picks_bans",
    cumple=pb[config.PB_HEROE].isin(heroes_validos),
    impacto="Un héroe desconocido no tiene nombre: no se puede mostrar ni comparar entre parches",
    responsable="Equipo de análisis (actualizar el catálogo de OpenDota)",
)

# El parche debe existir en el catálogo de parches
cal.evaluar(
    regla="patch existe en el catálogo de parches",
    dimension="Validez",
    tabla="main_metadata",
    cumple=meta["patch"].isin(parches["patch"]),
    impacto="Sin versión legible, la partida no entra en ningún grupo 'parche a parche'",
    responsable="Equipo de análisis (actualizar el catálogo de OpenDota)",
)

# Solo Captains Mode tiene draft de capitanes. Otro modo no es un dato "erróneo",
# sino una partida FUERA del dominio de la pregunta.
cal.evaluar(
    regla="modo de juego = Captains Mode",
    dimension="Validez",
    tabla="main_metadata",
    cumple=meta["game_mode"].eq(2),
    impacto="En otros modos (p. ej. All Pick) no hay bans de capitanes: sus picks no reflejan prioridades de draft",
    responsable="Equipo de análisis (excluir en Silver con regla crítica)",
)

# La fecha en texto debe poder convertirse a fecha real
cal.evaluar(
    regla="fecha convertible",
    dimension="Validez",
    tabla="main_metadata",
    cumple=meta["fecha"].notna(),
    impacto="Sin fecha no se puede ubicar la partida en un mes ni verificar su parche",
    responsable="Mantenedor del dataset",
)
````

### 8.4 Consistencia (aquí está el valor diferencial)

````python
# 1 — Integridad referencial: cada acción de draft pertenece a una partida conocida
cal.evaluar(
    regla="match_id de picks_bans existe en main_metadata",
    dimension="Consistencia",
    tabla="picks_bans ↔ main_metadata",
    cumple=pb[config.PB_PARTIDA].isin(meta["match_id"]),
    impacto="Acciones 'huérfanas' no tienen parche ni resultado: no pueden entrar al análisis",
    responsable="Mantenedor del dataset",
)

# 2 — Draft completo: 10 picks por partida, 5 por cada lado (team 0 = Radiant, 1 = Dire)
picks_lado = (pb[pb[config.PB_ES_PICK]]
              .groupby([config.PB_PARTIDA, config.PB_EQUIPO]).size()
              .unstack(fill_value=0))
draft_completo = picks_lado.get(0, 0).eq(5) & picks_lado.get(1, 0).eq(5)
cal.evaluar(
    regla="10 picks por partida (5 por lado)",
    dimension="Consistencia",
    tabla="main_metadata ↔ picks_bans",
    cumple=meta["match_id"].map(draft_completo),   # partidas sin draft → NaN → no cumple
    impacto="Un draft incompleto o ausente subestima el pick rate de los héroes de esa partida",
    responsable="Mantenedor del dataset / equipo (excluir en Silver)",
)

# 3 — Coherencia temporal: la fecha cae dentro de la carpeta de origen
#     ("2025" es anual → se compara el año; "2026MM" es mensual → año y mes)
es_anual = meta["periodo_origen"].str.len() == 4
cumple_periodo = pd.Series(
    np.where(es_anual,
             meta["fecha"].dt.strftime("%Y") == meta["periodo_origen"],
             meta["fecha"].dt.strftime("%Y%m") == meta["periodo_origen"]),
    index=meta.index)
cal.evaluar(
    regla="fecha dentro de su carpeta de origen",
    dimension="Consistencia",
    tabla="main_metadata",
    cumple=cumple_periodo,
    impacto="Una partida en la carpeta equivocada se cuenta en el mes incorrecto y puede duplicarse entre meses",
    responsable="Mantenedor del dataset",
)

# 4 — La partida se jugó DESPUÉS del lanzamiento de su parche
lanzamiento = meta["patch"].map(parches.set_index("patch")["lanzamiento"])
cal.evaluar(
    regla="fecha posterior al lanzamiento del parche",
    dimension="Consistencia",
    tabla="main_metadata ↔ catálogo de parches",
    cumple=meta["fecha"] >= lanzamiento,
    impacto="Una partida asignada a un parche que aún no existía distorsiona la comparación parche a parche",
    responsable="Mantenedor del dataset / OpenDota",
)

# 5 — Marcador vs. kills de los jugadores (solo 2026: players)
#     Radiant = player_slot 0–4 (< 128); Dire = 128–132.
#     OJO (conocimiento del dominio): si un héroe muere por una torre, un creep o un
#     monstruo neutral, la muerte SUMA al marcador del rival pero NO se atribuye a ningún
#     jugador. Por eso la regla correcta es "marcador >= suma de kills", no "=".
if players is not None:
    kills_rad = players[players["player_slot"] < 128].groupby("match_id")["kills"].sum()
    m26 = meta[meta["match_id"].isin(players["match_id"])]
    diferencia = m26["radiant_score"] - m26["match_id"].map(kills_rad)
    print("Diferencia (radiant_score − suma de kills), nº de partidas:")
    print(diferencia.value_counts().sort_index().to_string())
    cal.evaluar(
        regla="radiant_score >= suma de kills de Radiant (2026)",
        dimension="Consistencia",
        tabla="main_metadata ↔ players",
        cumple=diferencia.ge(0),
        impacto="Si los jugadores suman MÁS kills que el marcador, alguna de las dos tablas está mal registrada",
        responsable="Mantenedor del dataset",
    )
else:
    print("players no cargado: se omite la regla del marcador")
````

### 8.5 Plausibilidad

````python
# Umbrales justificados en la interpretación de la sección 8.8
DURACION_MIN, DURACION_MAX = 10, 120   # minutos

cal.evaluar(
    regla=f"duración entre {DURACION_MIN} y {DURACION_MAX} min",
    dimension="Plausibilidad",
    tabla="main_metadata",
    cumple=meta["duracion_min"].between(DURACION_MIN, DURACION_MAX),
    impacto="Partidas extremadamente cortas (rendición o abandono) o largas pueden no reflejar un draft jugado con normalidad",
    responsable="Equipo de análisis (marcar en Silver; no borrar)",
)

cal.evaluar(
    regla="human_players = 10",
    dimension="Plausibilidad",
    tabla="main_metadata",
    cumple=meta["human_players"].eq(10),
    impacto="Una partida profesional con menos de 10 jugadores no es comparable con el resto",
    responsable="Mantenedor del dataset",
)

cal.evaluar(
    regla="primera sangre entre 0 y la duración",
    dimension="Plausibilidad",
    tabla="main_metadata",
    cumple=meta["first_blood_time"].between(0, meta["duration"]),
    impacto="Un tiempo de primera sangre fuera de la partida indica un registro corrupto",
    responsable="Mantenedor del dataset",
)
````

### 8.6 Oportunidad

````python
print(f"Cobertura: {meta['fecha'].min():%Y-%m-%d} → {meta['fecha'].max():%Y-%m-%d}")
cobertura = (meta.groupby("parche", observed=True)
                 .agg(partidas=("match_id", "nunique"), desde=("fecha", "min"), hasta=("fecha", "max")))
display(cobertura)

# Regla 1: cada parche tiene volumen suficiente para comparar   (justificado en 8.8)
MIN_PARTIDAS_PARCHE = 1000
cal.evaluar(
    regla=f"parche con al menos {MIN_PARTIDAS_PARCHE} partidas",
    dimension="Oportunidad",
    tabla="main_metadata (por parche)",
    cumple=cobertura["partidas"].ge(MIN_PARTIDAS_PARCHE),
    impacto="Un parche con pocas partidas (p. ej. si solo cae al inicio o al final del período) da métricas poco estables",
    responsable="Equipo de análisis (advertirlo al comparar)",
)

# Regla 2: cada mes tiene un volumen comparable (≥ 50 % de la mediana mensual)
por_mes = meta.groupby("mes")["match_id"].nunique()
display(por_mes.to_frame("partidas").T)
cal.evaluar(
    regla="mes con al menos 50 % de la mediana mensual",
    dimension="Oportunidad",
    tabla="main_metadata (por mes)",
    cumple=por_mes.ge(0.5 * por_mes.median()),
    impacto="Un mes incompleto (p. ej. sep-2026, que llega solo al día 13) pesa poco y no representa el mes entero",
    responsable="Mantenedor del dataset (próxima actualización) / equipo (advertirlo)",
)
````

Una fila = una acción de draft, enriquecida con datos de la partida, el parche y el nombre del héroe.
Se excluyen **solo** los registros que incumplen reglas **críticas** explícitas; los rechazados se guardan para auditoría.

````python
draft_stage = (
    pb.rename(columns={config.PB_PARTIDA: "match_id", config.PB_ES_PICK: "es_pick",
                       config.PB_HEROE: "hero_id", config.PB_EQUIPO: "equipo_cod", config.PB_ORDEN: "orden"})
      [["match_id", "es_pick", "hero_id", "equipo_cod", "orden"]]
      .merge(meta[["match_id", "fecha", "patch", "parche", "radiant_win", "duracion_min", "leagueid", "game_mode"]],
             on="match_id", how="left", validate="many_to_one")
      .merge(heroes[["hero_id", "heroe"]], on="hero_id", how="left", validate="many_to_one")
)
draft_stage["es_pick"] = draft_stage["es_pick"].astype(bool)
draft_stage["equipo"] = np.where(draft_stage["equipo_cod"] == 0, "Radiant", "Dire")
draft_stage["gano"] = ((draft_stage["equipo"] == "Radiant") & draft_stage["radiant_win"].eq(True)) | \
                      ((draft_stage["equipo"] == "Dire") & draft_stage["radiant_win"].eq(False))

# Reglas críticas (justificadas en 10.1 con los resultados de la sección 8)
criticas = (
    draft_stage["radiant_win"].notna()     # la partida existe en main_metadata
    & draft_stage["heroe"].notna()         # el héroe existe en el catálogo
    & draft_stage["parche"].notna()       # se conoce el parche
    & draft_stage["game_mode"].eq(2)       # solo Captains Mode es comparable
)
rechazados = draft_stage.loc[~criticas]
draft = draft_stage.loc[criticas].copy()

rechazados.to_csv(config.OUTPUTS / "silver_rechazados.csv", index=False, encoding="utf-8-sig")
try:
    draft.to_parquet(config.SILVER / "draft_silver.parquet", index=False)
    formato = "Parquet"
except Exception:
    draft.to_csv(config.SILVER / "draft_silver.csv", index=False, encoding="utf-8-sig")
    formato = "CSV"
print(f"Silver: {len(draft):,} acciones válidas | {len(rechazados):,} rechazadas "
      f"({100 * len(rechazados) / len(draft_stage):.2f} %) | formato: {formato}")
print(f"Partidas en Silver: {draft['match_id'].nunique():,} | Parches: {draft['parche'].nunique()}")
````

### 12.3 Equilibrio entre lados: Radiant vs. Dire

````python
# % de victorias de Radiant por parche (50 % = lados equilibrados)
lados = (meta.groupby("parche", observed=True)
             .agg(partidas=("match_id", "nunique"),
                  radiant_win_pct=("radiant_win", lambda s: 100 * s.mean())))
# Margen de error al 95 % de una proporción: 1.96 * raíz(p * (1 - p) / n)
p = lados["radiant_win_pct"] / 100
lados["margen_95_pts"] = 100 * 1.96 * np.sqrt(p * (1 - p) / lados["partidas"])
display(lados.round(2))
print(f"Radiant en todo el período: {100 * meta['radiant_win'].mean():.1f} % de {len(meta):,} partidas")

fig, ax = plt.subplots(figsize=(10, 4.5))
x = lados.index.astype(str)
ax.plot(x, lados["radiant_win_pct"], marker="o", color=COLORES["steel"], label="Radiant")
ax.axhline(50, color=COLORES["gray"], ls="--", label="50 % (equilibrio)")
for xi, (pct, n) in zip(x, zip(lados["radiant_win_pct"], lados["partidas"])):
    ax.annotate(f"{pct:.1f} %\n(n={n:,})", (xi, pct), textcoords="offset points",
                xytext=(0, 8), ha="center", fontsize=8)
ax.set(title="% de victorias de Radiant por parche", xlabel="Parche", ylabel="% de victorias de Radiant",
       ylim=(min(45, lados["radiant_win_pct"].min() - 3), max(55, lados["radiant_win_pct"].max() + 4)))
ax.legend(loc="lower right")
plt.tight_layout()
plt.savefig(config.FIGURAS / "radiant_vs_dire_por_parche.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Interpretación**

**Entre héroes:** el parche **más equilibrado** fue **7.39**: tuvo la menor dispersión del win rate (2.9 puntos) y el **91 %** de los héroes con muestra suficiente ganó entre 45 % y 55 %. El **menos equilibrado** fue **7.37** (dispersión 4.7; 79 % en la franja 45–55 %), seguido de **7.41** (4.3; 78 %). En 7.41 conviven extremos, como muestra la tabla de héroes extremos: algunos superan el 60 % de victorias y otros quedan por debajo del 40 %.

**Muestra pequeña:** algunos extremos vienen de pocas partidas. En 7.37, el win rate más alto (Omniknight, 65.6 %) se basa en solo **32 picks**, y en 7.41 el más bajo (Elder Titan, 37.3 %) en 51. Por eso se exige un mínimo de 30 picks y estos casos se leen con cautela. Otros extremos sí tienen mucha muestra y son más confiables: en 7.40, Treant Protector ganó el **64.0 % en 756 picks**, y en 7.41 Lone Druid ganó el 60.7 % en 425. En 7.41 **los 127 héroes superan el mínimo**.

**Entre lados:** Radiant gana **más del 50 % en los 5 parches** (de 51.3 % a 52.6 %; 52.1 % en todo el período). El margen de error al 95 % va de ±0.7 a ±1.4 puntos. En 7.38, 7.39, 7.40 y 7.41 la ventaja supera ese margen; en 7.37 queda en el límite (51.3 ± 1.4).

**Para el analista:** el lado del mapa aporta una ventaja pequeña pero constante (~2 puntos). Además, en 7.41 hay más héroes fuera de balance que en 7.39, así que elegir o negar esos héroes pesa más en el parche actual. Esto **describe** lo ocurrido; no explica su causa.
